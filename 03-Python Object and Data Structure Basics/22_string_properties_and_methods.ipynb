**String Properties and Methods**

Immutability

Si escribimos name = "Sam", no podemos hacer name[0] = "P" para convertirlo en "Pam"

Tendriamos que hacer:

```python
name = "Sam"
last_letters = name[1:]
new_name = "P" + last_letters
new_name
```

Otras cosas que podemos hacer es concatenar textos:

```python
x = "Hello World"
x = x + " it is beautiful outside!"
x
```

```python
letter = "z"
letter * 10
```

```python
"2" + "3"
```

Tambien podemos hacer uso de los metodos propios de str:

```python
texto = "Hello World"

print(texto)
print(texto.upper())
print(texto.split())
```
