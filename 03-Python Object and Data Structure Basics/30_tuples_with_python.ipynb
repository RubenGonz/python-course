**Tuples with Python**

Son listas como las que hemos trabajado [ ] pero con la particularidad de que son inmutables y se escriben con ( )

```python
tuple = (1,2,3)
list = [1,2,3]
print(f"{tuple}:{type(tuple)}")
print(f"{list}:{type(list)}")
```

Otra diferencia es que solo cuentan con dos metodos:

```python
t = ("a", "c", "b", "a")
print(t.count("a"))
print(t.index("b"))
```
