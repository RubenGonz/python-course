**Dictionaries in Python**

Los diccionarios son listas pero indexadas por un string por lo que no se guian por un orden sino por un index que nosotros le ponemos:

```python
edades={"Marta":25,"Juan":30,"Ana":22}
print(f"Edades: {edades}")
print(f"Juan: {edades["Juan"]}")
```

Buscar a traves de ellos con:

```python
d = {"k1": 123, "k2": [0, 1, 2], "k3": {"insideKey": 100}}
print(d["k1"])
print(d["k2"][1])
print(d["k3"]["insideKey"])
```

O modificar su valor con:

```python
original = {"first": 1, "second": 2}
original["third"] = 3
print(original)
original["second"] = "changed"
print(original)
```

Y mirar sus valores con:

```python
dictionary = {"key1": "value1", "key2": "value2"}
print(dictionary.keys())
print(dictionary.values())
print(dictionary.items())
```
