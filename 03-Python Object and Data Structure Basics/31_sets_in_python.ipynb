**Sets in Python**

Los sets tambien son grupos de elementos que no se pueden repetir y se reconocen por { }

```python
myset = set()
myset.add(1)
myset.add(2)
myset
```

Al intentar volver a añadir un valor ya existente, simplemente no lo haría:

```python
myset.add(2)
myset
```

Esta funcionalidad es muy util para casos como este:

```python
set([1,1,1,2,2,3,4,4,5,"a","b","b"])
```
