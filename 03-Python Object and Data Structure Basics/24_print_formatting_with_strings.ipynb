**Print Formatting with Strings**

El metodo .format() permite hacer acciones como:

```python
print("El texto: {} fue insertado".format("'HOLA'"))
```

```python
print("La casa de {} es {} y {}".format("Pedro", "bonita", "roja"))
```

```python
print("La casa de {1} es {2} y {0}".format("roja", "Pedro", "bonita"))
```

```python
print("La casa de {nombre} es {belleza} y {color}".format(color="roja",nombre="Pedro",belleza="bonita"))
```

```python
color="roja"
nombre="Pedro"
belleza="bonita"

print("La casa de {nombre} es {belleza} y {color}".format(color =color,nombre=nombre,belleza=belleza))
```

```python
resultado = 100 / 777
print("El resultado es: {}".format(resultado))
print("El resultado es: {:.3f}".format(resultado))
```

Otra forma de hacerlo es con:

```python
color="roja"
nombre="Pedro"
belleza="bonita"

print("La casa de {nombre} es {belleza} y {color}".format(color =color,nombre=nombre,belleza=belleza))
print(f"La casa de {nombre} es {belleza} y {color}")
```
