**Indexing and Slicing with Strings**

Se puedes tratar los strings en python como arrays gracias a: [ ]

Para intexar:

```python
texto = "Hello World"
texto
```

```python
texto[1]
```

```python
texto[-1]
```

Para hacer un slice:

```python
texto = "abcdefghijk"
texto
```

[number:] para ver todo el contenido que hay despues de ese index

```python
texto[2:]
```

[:number] para ver todo el contenido hasta ese index

```python
texto[:3]
```

[number:number] para ver todo el contenido entre esos index

```python
texto[3:6]
```

[::number] para ver el contenido si se hicieran saltos

```python
texto[::2]
```

Por lo que tambien podriamos hacer:

```python
texto[::-1]
```
