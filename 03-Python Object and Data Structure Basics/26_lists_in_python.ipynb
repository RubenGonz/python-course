**Lists in Python**

Para definir lista lo podemos hacer como:

```python
lista = ["String", 23, 45.67, True]
lista
```

Y usar las funciones como en los str:

```python
print(len(lista))
print(lista[1])
print(lista[1:-1])
```

Y tambien tenemos otras funciones como:

```python
nueva_lista = lista + ["Nuevo elemento"]
print(lista)
print(nueva_lista)
```

Una diferencia con los str es que las listas si son mutables:

```python
lista[0] = "String cambiado"
lista
```

Y tienen nuevos metodos como:

```python
print(f"Lista original:          {lista}")

lista.append("Nuevo elemento al final")
print(f"Lista despues de append: {lista}")

lista.pop(0)
print(f"Lista despues de pop:    {lista}")
```

```python
num_list = [2,10,5,3,8,4]
print(f"Antes:              {num_list}")
num_list.sort()
print(f"Despues de sort:    {num_list}")
num_list.reverse()
print(f"Despues de reverse: {num_list}")
```
