**I/O with Basic Files in Python**

Esta es una seccion que se ampliara en el futuro y trata del manejejo de archivos, empezando con el .txt

Donde con esto podremos crear y escribir txt:

```python
%%writefile ./files/myfile.txt
Hello this is a text file.
And this is the second line.
```

Tambien podemos ver lo que hemos escrito con:

```python
my_file = open('./files/myfile.txt')
print(my_file.read())
```

Si lo volvemos a intentar fallara y esto se debe a que el cursos termina al final del archivo y tenemos que volver a colorcarlo al principio:

```python
print(f"Texto:\n{my_file.read()}")
my_file.seek(0)
print(f"Texto tras el seek:\n{my_file.read()}")
```

También tenemos mas metodos como .readlines() que dividen en una lista las lineas que tenemos:

```python
my_file.seek(0)
my_file.readlines()
```

Tambien es importante decir que mientras estemos haciendo uso de ese archivo no vamos a poder hacer acciones como eliminarlo por lo que para ello tenemos que tener un control y saber tambien de la función:

```python
my_file.close()
```

Para no estar pendientes de esto tambien podemos trabajar con with que nos libra de este problema

```python
with open('./files/myfile.txt') as my_file:
    contents = my_file.read()
    print(contents)
```

Tambien podemos hacer uso de permisos para los archivos con el atributo mode:

- r: read-only
- r+: read and write
- w: write only
- w+: write and read
- a: append only
- a+: append and read

```python
with open('./files/myfile.txt', mode='a+') as my_file:
    my_file.write('\nThis is the new line added.')
    my_file.write('\nThis is another new line added.')
    my_file.seek(0)        
    print(my_file.read())  
```

Esta funcion tambien sirve para crear archivos:

```python
with open('./files/my_new_file.txt', mode='w+') as my_new_file:
    my_new_file.write('This is a new file created using Python.\n')  
    my_new_file.seek(0)    
    print(my_new_file.read())  
```
